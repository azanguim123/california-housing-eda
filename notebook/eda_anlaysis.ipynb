# __Setuyp & Load__

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)
sns.set_theme(style="whitegrid")

# Load once, keep a pristine copy
df_raw = pd.read_csv("../data/house.csv")  #original -never touched again
df = df_raw.copy()                          #working copy for all cleanig steps

# First look
print("Shape:", df.shape)
df.head()
df.info()
df.describe()
```

# __ Cleaning __

```python
# Missing values
print(df.isnull().sum())

# Duplicates
n_dupes = df.duplicated().sum()
print(f"duplicates: {n_dupes}")
df = df.drop_duplicates().reset_index(drop=True)
print("Shape after: ", df.shape)

#Dtypes
print(df.dtypes)
```
